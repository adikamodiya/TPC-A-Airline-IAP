# Final Business Challenge - Explanation & Plan

The FinalBusiness Challenge asks one question: **How is our booking business performing, and where should management focus its attention?** This cell explains what the challenge demands, how to handle it, and a ready-to-use draft mapped to our actual Q01-Q20 results.

---

## 1. What the challenge wants

Use the full pipeline output - Gold datasets, SQL analysis, Dashboard, and Genie - and present, **in the final cell of the SQL notebook**, exactly:

| # | Deliverable | Count | Must include |
|---|-------------|-------|--------------|
| 1 | Key findings | 3 | A specific query result + metric value for each |
| 2 | Opportunities (revenue growth or operational improvement) | 2 | Data evidence |
| 3 | Risks / areas needing further investigation | 2 | Data evidence |
| 4 | Recommended management action | 1 per item (7 total) | One for every finding, opportunity, and risk |

**Guardrails (graded):**
- Clearly **separate what the data directly supports** from any assumption or hypothesis.
- **No real-world causation claims** - this is synthetic practice data (e.g. do not say revenue grew *because of* marketing).
- **Label every cited metric with its source query** - e.g. Q03 shows annual revenue of X.

## 2. How to handle it

1. **Do not write new queries.** Q01-Q20 in the SQL notebook already contain title, interpretation, and a management action in each comment header - the summary is a curation exercise, not new analysis.
2. **Use a fixed formula per item:** Finding -> Evidence (Qxx shows metric = value) -> Action. Seven items, same pattern, nothing vague.
3. **Pick findings that are quantified and decision-relevant** (growth trend, concentration, refunds, geography flows) - not restatements of raw tables.
4. **Split tone:** state data-backed conclusions as facts; anything speculative gets prefixed "Hypothesis (not proven by data): ...".
5. **Placement matters:** the deliverable must be the **last cell of the SQL notebook** - copy the final draft there once reviewed.

## 3. Draft summary (mapped to our verified Q results)

### Key findings (data-supported)
- **F1 - Revenue dipped in 1995, then recovered and grew every year.** Q03 shows net revenue fell **-5.83% in 1995 ($8.05M)**, then rose +4.31% (1996), +1.07% (1997), +4.19% (1998), reaching **$8.85M in 1998**. *Action:* trigger a pricing/loyalty review if any year turns negative again.
- **F2 - Revenue is well diversified, no customer concentration.** Q06 shows the top-10 passengers contribute only **1.86%** of the **$42.34M** total (largest single customer: Michael Ali, $113.6K); Q11 shows the 10 airlines each hold ~10% share. *Action:* maintain the broad loyalty program rather than over-investing in individual VIP handling.
- **F3 - Booking economics are healthy at scale.** Q01 shows **$42.34M net revenue from 7,618 bookings and 1,493 active passengers**, averaging **$5,558.42 per booking** (1994-1998). *Action:* use this as the baseline scorecard for future periods.

### Opportunities (evidence-based)
- **O1 - Business travel is the largest revenue segment.** Q04 shows BUSINESS = **$9.08M (21.44% of revenue**, top of 5 purposes). *Action:* steer fare products and cabin mix toward business travelers.
- **O2 - Geographic revenue flows are imbalanced.** Q20 shows **EUROPE is the biggest inbound region (+$727K** passenger-vs-airline flow) while **AMERICA is the most outbound (-$625K)**. *Action:* investigate route/pricing mix on America-based routes to capture more local revenue.

### Risks / investigation areas (evidence-based)
- **R1 - Refund exposure is material.** Q14 shows refund-flagged lines account for **9.88% of net revenue**. *Action:* audit refund drivers (status, priority, agent) before setting refund policy.
- **R2 - Flight-duration data appears degenerate.** Q16 shows all flights arrive the **same day (duration 0 days)**, which limits operational analysis. *Hypothesis (not proven by data):* this may be a source-data artifact rather than real operations. *Action:* investigate bronze/silver ingestion of arrival dates before using duration in decisions.
