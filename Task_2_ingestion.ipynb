```python
catalog_name = dbutils.widgets.get("catalog_name")
pipeline_id = dbutils.widgets.get("pipeline_id")

spark.sql(f"""
create or replace table {catalog_name}.bronze.passengers as
select *, _metadata.file_path as source_file, current_timestamp() as _ingested_at , '{pipeline_id}' as pipeline_run_id   from
  read_files(
    '/Volumes/{catalog_name}/landing/raw/passengers/',
    format => 'csv',
    header => true,
    schemaEvolutionMode => 'addNewColumns',
    inferColumnTypes=> false
  )
""")

```
